```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/data.csv')
```

```python
data['TransactionStartTime'] = pd.to_datetime(data['TransactionStartTime'])
numerical_columns = ["Amount", "Value", "CountryCode", "PricingStrategy", "FraudResult"]
numerical_summary = data[numerical_columns].describe()

numerical_summary.loc["range"] = numerical_summary.loc["max"] - numerical_summary.loc["min"]
numerical_summary.loc["skew"] = data[numerical_columns].skew()
numerical_summary.loc["kurtosis"] = data[numerical_columns].kurt()

categorical_columns = ["CurrencyCode", "ProviderId", "ProductId", "ProductCategory", "ChannelId"]
categorical_summary = {}
for col in categorical_columns:
    categorical_summary[col] = data[col].value_counts()

print("Numerical Summary Statistics:\n", numerical_summary)
print("\nCategorical Summary Statistics:")
for col, counts in categorical_summary.items():
    print(f"{col}:\n{counts}\n")
```

```python
numerical_columns = ["Amount", "Value", "CountryCode", "PricingStrategy", "FraudResult"]
sns.set(style="whitegrid")

for col in numerical_columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(data[col], kde=True, bins=30, color='blue')
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
numerical_columns = ["Amount", "Value", "CountryCode", "PricingStrategy", "FraudResult"]
correlation_matrix = data[numerical_columns].corr()
correlation_matrix
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Numerical Features")
plt.show()

sns.pairplot(data[numerical_columns], diag_kind="kde", corner=True, hue="FraudResult")
plt.suptitle("Pair Plot of Numerical Features", y=1.02)
plt.show()

# Scatter plots with regression lines for key pairs
key_pairs = [("Amount", "Value"), ("Amount", "FraudResult")]
for x, y in key_pairs:
    plt.figure(figsize=(8, 6))
    sns.regplot(x=data[x], y=data[y], scatter_kws={"alpha": 0.3}, line_kws={"color": "red"})
    plt.title(f"Scatter Plot of {x} vs {y}")
    plt.xlabel(x)
    plt.ylabel(y)
    plt.show()
```

```python
# Check for explicit missing (NaN) and '0' values
missing_summary = {}

for col in data.columns:
    if data[col].dtype != 'object':  # Numerical columns
        missing_summary[col] = {
            "Missing (NaN)": data[col].isnull().sum(),
            "Zero Values": (data[col] == 0).sum(),
            "Total Missing": data[col].isnull().sum() + (data[col] == 0).sum(),
            "Percentage Missing (%)": ((data[col].isnull().sum() + (data[col] == 0).sum()) / len(data)) * 100,
        }
    else:  # Categorical columns
        missing_summary[col] = {
            "Missing (NaN)": data[col].isnull().sum(),
            "Zero Values": 0,  # No '0' in categorical columns
            "Total Missing": data[col].isnull().sum(),
            "Percentage Missing (%)": (data[col].isnull().sum() / len(data)) * 100,
        }

# Convert to DataFrame for easier interpretation
missing_df = pd.DataFrame(missing_summary).T
missing_df = missing_df.sort_values(by="Total Missing", ascending=False)

print("Summary of Missing Values (Including 0):")
print(missing_df)
```
